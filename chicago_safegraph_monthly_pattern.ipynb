```python
directory = 'E:/Data/Safegraph_monthly_pattern/'

# iterate over files in that directory
for filename in os.listdir(directory):
    # checking if it is a file
    if filename.startswith("monthly_pattern-"): # for files start with a prefix #
        f = os.path.join(directory, filename)
        df = pd.read_csv(f, on_bad_lines='skip', encoding = "ISO-8859-1")
        df = df[df['city'] == 'Chicago']
        df = df[['opened_on', 'closed_on', 'tracking_closed_since', 'top_category', 'raw_visitor_counts', 'distance_from_home', 'raw_visit_counts', 
                 'median_dwell', 'sub_category', 'naics_code', 'location_name', 'visitor_home_cbgs', 'date_range_start', 'date_range_end', 
                 'city', 'region', 'poi_cbg', 'iso_country_code']]         
        df = df.assign(visitor_home_cbgs=df.visitor_home_cbgs.str.split(",")).explode('visitor_home_cbgs')
        df = df.replace({'{':'', '}':''}, regex=True)
        df[['home_cbgs','count']] = df['visitor_home_cbgs'].str.split('":',expand=True)
        df = df.replace({'"': ''}, regex=True)
        df['date_range_start'] = df['date_range_start'].str[:10] # work with the date #
        df['date_range_end'] = df['date_range_end'].str[:10] # work with the date #
        df = df[df['count'].notna()] # get rid of missing value #
        df = df.groupby(['top_category', 'sub_category', 'naics_code', 'date_range_start', 'date_range_end', 'city', 'region', 'poi_cbg', 'iso_country_code', 
                         'home_cbgs']).agg({'count': 'sum'}) # aggregate data #   
        #df = df.groupby(['top_category', 'sub_category', 'naics_code', 'date_range_start', 'date_range_end', 'city', 'region', 'poi_cbg', 'iso_country_code', 'home_cbgs']).agg({'count': 'sum', 'raw_visitor_counts': 'mean', 'distance_from_home': 'mean', 'median_dwell': 'mean'}) # aggregate
        df.to_csv('E:/chicago_crime_project/data/safegraph_monthly_pattern/{}'.format(filename)) 
        
        
        ### Test
import numpy as np
directory = 'E:/Data/Safegraph_monthly_pattern/'

# iterate over files in that directory
for filename in os.listdir(directory):
    # checking if it is a file
    if filename.startswith("monthly_pattern-"): # for files start with a prefix #
        f = os.path.join(directory, filename)
        df = pd.read_csv(f, on_bad_lines='skip', encoding = "ISO-8859-1")
        df = df[df['city'] == 'Chicago']
        df = df[['opened_on', 'closed_on', 'tracking_closed_since', 'top_category', 'raw_visitor_counts', 'distance_from_home', 'raw_visit_counts', 
                 'median_dwell', 'sub_category', 'naics_code', 'location_name', 'visitor_home_cbgs', 'date_range_start', 'date_range_end', 
                 'city', 'region', 'poi_cbg', 'iso_country_code']]         
        df['date_range_start'] = df['date_range_start'].str[:10] # work with the date #
        df['date_range_end'] = df['date_range_end'].str[:10] # work with the date #
        
        # Define a lambda function to compute the weighted mean:
        wm = lambda x: np.average(x, weights=df.loc[x.index, "raw_visitor_counts"])

        # Define a dictionary with the functions to apply for a given column:
        # the following is deprecated since pandas 0.20:
        # f = {'adjusted_lots': ['sum'], 'price': {'weighted_mean' : wm} }
        # df.groupby(["contract", "month", "year", "buys"]).agg(f)

        # Groupby and aggregate with namedAgg [1]:
        df = df.groupby(['top_category', 'sub_category', 'naics_code', 'date_range_start', 'date_range_end', 'city', 'region', 'poi_cbg', 'iso_country_code']).agg(raw_visitor_counts=("raw_visitor_counts", "sum"), 
                                                                   distance_from_home=("distance_from_home", wm),
                                                                    median_dwell=("median_dwell", wm))
        df.to_csv('E:/chicago_crime_project/data/safegraph_monthly_pattern/weighted_mean/{}'.format(filename)) 
```
