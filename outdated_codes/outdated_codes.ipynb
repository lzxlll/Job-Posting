### The new (live) branches within a predetermined radius with the parent firm (origin).

- Add a new column in `dfBranch2` that contains the distance between `城市_city` and `parent_city` by looking up the distance from your distance matrix.

- Filter the rows in `dfBranch2` where the distance is less than or equal to 200, 300, 400, 500, 1000 km.

- Group the filtered dataframe by `注册年份_estiblishYear` and `parent_city` and count the number of observations.

```python
# Load the distance matrix generated in the previous steps
distance_matrix = pd.read_csv('D:/Dropbox/Dropbox/Paper with Yao-yu/Spatial Data Programming/Data/distance_matrix.csv', header=0, index_col=0)

# Step 1: Add a new column with the distance between the two cities to 'dfBranch2' 
dfBranch2['distance'] = dfBranch2.apply(lambda row: distance_matrix.loc[row['城市_city'], row['parent_city']], axis=1)

# Step 2: Filter rows where the distance is less than or equal to 200, 300, 400, 500, 1000 km. Then, group by year and parent city and count the number of observations
# Define distance intervals
distances_intervals = [(0, 200), (200, 300), (300, 400), (400, 500), (500, 1000), (1000, float('inf'))]
results = {}

for dist_interval in distances_intervals:
    # Filter rows where the distance is within the current threshold interval
    df_filtered = dfBranch2[(dfBranch2['distance'] >= dist_interval[0]) & (dfBranch2['distance'] < dist_interval[1])]

    # Group by year and parent city and count the number of observations
    result = df_filtered.groupby(['注册年份_estiblishYear', 'parent_province', 'parent_city'])['公司ID'].nunique().reset_index(name='branchCount')

    # Store the result in the dictionary using the distance interval as the key
    results[dist_interval] = result


# Initialize the final result with the first dataframe
dfBranch4 = results[(0,200)].rename(columns={'branchCount': 'branchCount_Near200pct'})

# renames the 'counts' column to 'branchCount_Near{dist}pct' for each dataframe before merging,
# where {dist} is the string representation of the distance interval

for dist_interval in distances_intervals[1:]:
    dist_interval_str = f"{dist_interval[0]}_{dist_interval[1]}" if dist_interval[1] != float('inf') else f"{dist_interval[0]}_inf"
    # Merge the current result with the final result
    dfBranch4 = dfBranch4.merge(results[dist_interval].rename(columns={'branchCount': f'branchCount_Near{dist_interval_str}pct'}), 
                                on=['注册年份_estiblishYear', 'parent_province', 'parent_city'], how='outer')

# rename '注册年份_estiblishYear' to 'year', rename 'parent_province' to '省份_province', rename 'parent_city' to '城市_city', rename 'counts' to 'branchCount_Near200pct'
dfBranch4.rename(columns={'注册年份_estiblishYear': 'year', 'parent_province': '省份_province', 'parent_city': '城市_city'}, inplace=True)

# keep '注册年份_estiblishYear' between 2017 and 2020 in 'result'
dfBranch4 = dfBranch4[(dfBranch4['year'] >= 2017) & (dfBranch4['year'] <= 2019)]
dfBranch4
```

```python
import pandas as pd
import numpy as np
from linearmodels.panel import PanelOLS
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import matplotlib

#matplotlib.rcParams['font.sans-serif'] = ['SimHei']  # 'SimHei' is a font that supports Chinese characters
#matplotlib.rcParams['axes.unicode_minus'] = False  # Avoids problems with rendering negative numbers

# Load data
df = pd.read_csv('D:/Dropbox/Dropbox/Paper with Yao-yu/Spatial Data Programming/Data/dfBranch_master2.csv')

translations = {
    '交通运输、仓储和邮政业': 'Transportation, Storage, and Postal Service',
    '住宿和餐饮业': 'Accommodation and Food Services',
    '信息传输、软件和信息技术服务业': 'Information Transmission, Software and Information Technology Services',
    '制造业': 'Manufacturing',
    '卫生和社会工作': 'Health and Social Work',
    '居民服务、修理和其他服务业': 'Residential Services, Repair, and Other Services',
    '建筑业': 'Construction',
    
    '房地产业': 'Real Estate',
    '批发和零售业': 'Wholesale and Retail Trade',
    '教育': 'Education',
    '文化、体育和娱乐业': 'Culture, Sports, and Entertainment',
    '水利、环境和公共设施管理业': 'Water Conservancy, Environment and Public Facilities Management',
    '电力、热力、燃气及水生产和供应业': 'Electricity, Heat, Gas, and Water Production and Supply',
    '科学研究和技术服务业': 'Scientific Research and Technical Services',
    '租赁和商务服务业': 'Leasing and Business Services',
    '金融业': 'Finance'
}
df['一级行业_categoryStrBig'] = df['一级行业_categoryStrBig'].map(translations)

# Generate dummies
df['year_t'] = pd.factorize(df['year'])[0] + 1
df['prov_id'] = pd.factorize(df['省份_province'])[0] + 1
df['city_id'] = pd.factorize(df['城市_city'])[0] + 1

# Rename variables '地区生产总值（当年价格（万元））', '工业企业数（个）', '专利授权数(件)', '发明授权数(件)' to 'gdp', 'indust', 'patent', 'invention'
df.rename(columns={'年末户籍人口（万人）': 'cityPop', '地区生产总值（当年价格（万元））': 'gdp', '工业企业数（个）': 'indust', '专利授权数(件)': 'patent', '发明授权数(件)': 'invention'}, inplace=True)

# List of variables
x_vars = ['gdp', 'indust', 'patent', 'invention']

# Replace missing values with 0 and take natural log for variables
for i in ['progShare_pct', 'branchCount_sector'] + x_vars:
    df[i].fillna(0, inplace=True)
    #df[i] = np.log(df[i] + 0.0001)

# Get all unique categories
categories = df['一级行业_categoryStrBig'].unique()

# List to store results
results = []

# Loop over categories and run regression
for category in categories:
    temp_df = df[df['一级行业_categoryStrBig'] == category].copy()
    temp_df.set_index(['year', 'city_id'], inplace=True)
    mod = PanelOLS.from_formula('branchCount_sector ~ progShare_pct + ' + ' + '.join(x_vars) + ' + C(prov_id) + C(year_t)', weights=temp_df['cityPop'], data=temp_df)
    result = mod.fit(cov_type='clustered', cluster_entity=True)
    results.append(result)


# Define a list to store categories and corresponding coefficients
category_coef = []

# Loop over the categories and results, and store the category and corresponding coefficient
for i, res in enumerate(results):
    category_coef.append((categories[i], res.params['progShare_pct']))

# Sort the list by the coefficient
category_coef.sort(key=lambda x: x[1])

# Create separate lists for sorted categories and coefficients
categories_sorted, coefs_sorted = zip(*category_coef)
```

```python
# Define a list to store categories, corresponding coefficients and confidence intervals
category_coef_ci = []

# Loop over the categories and results, and store the category, corresponding coefficient and confidence interval
for i, res in enumerate(results):
    ci = res.conf_int().loc['progShare_pct'].values
    category_coef_ci.append((categories[i], res.params['progShare_pct'], ci))

# Sort the list by the coefficient
category_coef_ci.sort(key=lambda x: x[1])

# Create separate lists for sorted categories, coefficients and confidence intervals
categories_sorted, coefs_sorted, ci_sorted = zip(*category_coef_ci)

# Get lower and upper bounds of confidence intervals
lower_bounds = [coef - ci[0] for coef, ci in zip(coefs_sorted, ci_sorted)]
upper_bounds = [ci[1] - coef for coef, ci in zip(coefs_sorted, ci_sorted)]
```

```python
import matplotlib.patches as mpatches

# Define colors
colors = plt.cm.viridis(np.linspace(0, 1, len(categories_sorted)))

# Now let's plot the coefficients with their confidence intervals
fig, ax = plt.subplots(figsize=(12, 7))

# Create patches for the legend
patches = [mpatches.Patch(color=colors[i], label=category) for i, category in enumerate(categories_sorted)]

# Plot each category with different color
for i, (coef, lower, upper) in enumerate(zip(coefs_sorted, lower_bounds, upper_bounds)):
    ax.errorbar(i, coef, yerr=[[lower], [upper]], fmt='o', color=colors[i], capsize=5, markersize=10)

# Add horizontal line at y = 0.279
ax.axhline(y=0.279, color='r', linestyle='--')

# Add shaded confidence interval between y = [.2562168, .3007975]
ax.fill_between(range(len(categories_sorted)), 0.2562168, 0.3007975, color='red', alpha=0.2)

# Add legend below the plot with multiple columns
box = ax.get_position()
ax.set_position([box.x0, box.y0, box.width, box.height])  # Adjust height to make space for the legend
ax.legend(handles=patches, loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2)  # Set bbox_to_anchor to position legend below the plot

ax.set_xlabel('Industrial Sectors (Major)', fontsize=14)
ax.set_ylabel('Coefficient and 95% Confidence Interval', fontsize=14)
ax.set_title('Coefficient of Data Programming by Sectors', fontsize=16)
ax.set_xticks(range(len(categories_sorted)))  # Set xticks to be indices
ax.set_xticklabels([])  # Remove xticklabels

ax.grid(True, which='both', axis='both', linestyle='--', linewidth=0.5)

plt.tight_layout()
plt.show()
```

### Determine the local firm's data programming tech upgrading path

- Define 5 occupation groups based on the `Ceneus occupation classification`.

```python
# Make sure 'soc_code' is numeric
dfLocal['soc_code'] = pd.to_numeric(dfLocal['soc_code'], errors='coerce')

# Define the bins and labels
bins = [(110000, 310000), (310000, 410000), (410000, 450000), (450000, 510000), (510000, float('inf'))]
labels = ['Business/Science', 'Service', 
          'Sales/Office', 'Construction/Maintenance', 
          'Production/Transportation']

# Create a function to do the classification
def classify_soc_code(code):
    if pd.isna(code):
        return np.nan
    for bin, label in zip(bins, labels):
        if code >= bin[0] and code < bin[1]:
            return label
    return 'Other'

# Apply the function to the 'soc_code' column to create the new column
dfLocal['socGroup'] = dfLocal['soc_code'].apply(classify_soc_code)

# print unique values with frequency in 'socGroup' column
print(dfLocal['socGroup'].value_counts())
```

- generate data programming share by occupation group by year-city.
- generate job posting share by occupation group by year-city.

```python
# Step 1: Aggregate the data
df_grouped = dfLocal.groupby(['year', '省份_province', '城市_city', 'socGroup']).agg({'posting_count':'sum', 'software_data':'mean'}).reset_index()

# rename 'posting_count' to 'localPostingCount_occ',  'software_data' to 'localProg_occ' in 'df_grouped'
df_grouped.rename(columns={'posting_count': 'localPostingCount_occ', 'software_data': 'localProg_occ'}, inplace=True)

# Compute the total 'localPostingCount_occ' for each 'year', '省份_province', '城市_city' combination
total_postings = df_grouped.groupby(['year', '省份_province', '城市_city'])['localPostingCount_occ'].transform('sum')

# Divide 'localPostingCount_occ' by the computed total and assign the result to a new column
df_grouped['posting_share'] = df_grouped['localPostingCount_occ'] / total_postings

# Step 2: Replace the average of 'software_data' to null if  the total of 'posting_count' is less than 2000
df_grouped['localProg_occ'] = df_grouped.apply(lambda row: None if row['localPostingCount_occ'] < 100 else row['localProg_occ'], axis=1)
df_grouped['posting_share'] = df_grouped.apply(lambda row: None if row['localPostingCount_occ'] < 100 else row['posting_share'], axis=1)

# Print the result
print(df_grouped)
```
