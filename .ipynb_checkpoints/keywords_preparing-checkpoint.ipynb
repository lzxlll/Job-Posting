```python
# List of technologies
technologies = ['Linux', 'Microsoft Excel', 'Atlassian JIRA']

# Generate similar software names for each technology
similar_software = []
for tech in technologies:
    prompt = f"Software similar to {tech} used in China"
    response = openai.ChatCompletion.create(
  model="gpt-3.5-turbo",
  messages=[
        {"role": "system", "content": "Software counterparts in China with simlar purpose"},
        {"role": "user", "content": "Any software used in China is similar to 'Adobe Systems Adobe Acrobat'? Please only show me the software names (only in Chinese) as the answer."}
    ]
)
    if response.choices[0].message.content:
        software_list = [line.split('. ', 1)[1] for line in response.choices[0].message.content.split('\n')t]
        similar_software.extend(software_list)
    else:
        similar_software.append('')

# Create a DataFrame
df = pd.DataFrame({
    'Technology': technologies * len(similar_software),
    'Software in China': similar_software * len(technologies)
})

# Print the DataFrame
print(df)
```
