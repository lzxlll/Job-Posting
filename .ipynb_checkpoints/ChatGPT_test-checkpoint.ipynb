```python
import os
import openai
import pandas as pd
import re
import os
from glob import glob
from datetime import datetime
import gc
import requests
import json
import time
```

```python
def classify_job_title(technologies, api_key):
    url = 'https://api.openai.com/v1/chat/completions'
    headers = {'Content-Type': 'application/json',
               'Authorization': f'Bearer {api_key}'}
    data = {'model': 'gpt-3.5-turbo-0301',
            'messages':[
                {
                'role': 'user', 
                'content': f'Please tell me "{technologies}" that are used in China. Only show software names (in Chinese) as the answer.'
                }
                       ]
            }
    response = requests.post(url, headers=headers, json=data)
    response_data = json.loads(response.text)
    if response.status_code != 200:
        raise Exception(response_data['error']['message'])
    return response_data['choices'][0]['message']['content']
```

```python
technologies = ['Data base user interface and query software', 'Accounting software']
soc_codes = []
api_key = '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'
start_time = time.time()
for technology in technologies:
    soc_code = classify_job_title(technology, api_key)
    soc_codes.append(soc_code)
print("--- %s seconds ---" % (time.time() - start_time))
```

```python
soc_codes
```

```python
sublists = []
for string in soc_codes:
    sublists += string.split('\n\n')

software_names = []
for sublist in sublists:
    for name in sublist.split('\n'):
        if len(name) > 0 and name[0].isdigit():
            name_parts = name.split('. ')
            if len(name_parts) >= 2:
                software_names.append(name_parts[1])

print(software_names)
```
