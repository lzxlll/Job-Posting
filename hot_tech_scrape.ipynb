```python
import requests
import pandas as pd
import re
```

```python
# We download the hot technologies with the usage from ONET OnLine: https://www.onetonline.org/search/hot_tech/
# The usage of each technology comes from: https://www.onetonline.org/search/hot_tech/example?e=Microsoft%20Excel
# Source: Lightcast external site job postings data for the US nationwide between July 1, 2021 and June 30, 2022. 
# “Percentage” represents the ratio of unique postings which mention the skill to all unique postings linked to the O*NET-SOC occupation.

# The following code is a scrape data code
df = pd.read_csv('E:/job_posting_project/data/tech_frequency/Hot_Technologies.csv')
software_names = df["Hot Technology"].tolist()
software_names
```

```python
# Loop through the list of software names and generate the corresponding URLs, generate a list
urls = [f"https://www.onetonline.org/search/hot_tech/example/{software_name.replace(' ', '_')}.csv?e={software_name.replace(' ', '%20')}&fmt=csv" for software_name in software_names]
```

```python
# Loop through the list of URLs and download each CSV file
for url in urls:
    # Send a GET request to the URL to download the CSV file
    response = requests.get(url)

    # Check if the request was successful
    if response.status_code == 200:
        # Extract the file name from the URL and save the CSV file to a local file
        # uses the re (regular expression) module in Python to extract a filename from a URL string
        # url.split("?")[0]: This code splits the URL string at the ? character and returns the first part of the string, which is the URL without any query parameters.
        # .split("/")[-1]: This code splits the URL string at the / character and returns the last part of the string, which is typically the filename.
        # re.sub('[^0-9a-zA-Z._-]+', '', ...): This code removes any characters in the filename that are not alphanumeric, dots, underscores, or hyphens. The re.sub() function in Python uses regular expressions to replace substrings in a string. In this case, it replaces any sequence of one or more characters that are not in the set [0-9a-zA-Z._-] with an empty string ''.
        filename = re.sub('[^0-9a-zA-Z._-]+', '', url.split("?")[0].split("/")[-1])
        filepath = "E:/job_posting_project/data/tech_frequency/" + filename
        with open(filepath, "wb") as file:
            file.write(response.content)
            print(f"File {filename} downloaded successfully!")
    else:
        print(f"Failed to download file from {url}.")
```

```python
import os
import pandas as pd

# Get a list of all CSV files in the directory
csv_files = [file for file in os.listdir('E:/job_posting_project/data/tech_frequency/') if file.endswith('.csv')]

os.chdir("E:/job_posting_project/data/tech_frequency") 

# Loop over each file and read it using pandas
dfs = []
for file in csv_files:
    # Read the CSV file
    df = pd.read_csv(file)
    
    # Create a new column named "Tech" and fill it with the name of the file (with underscores replaced with spaces)
    tech_name = file.replace('_', ' ')
    tech_name = tech_name.replace('.csv', '')
    df['Tech'] = tech_name
    
    # Append the updated dataframe to the list of dataframes
    dfs.append(df)

# Concatenate all the dataframes in the list into a single dataframe
merged_df = pd.concat(dfs, ignore_index=True)

# Save the merged dataframe to a new file
merged_df.to_csv('E:/job_posting_project/data/technology_skill_onet/tech_used_states.csv', index=False)
```
